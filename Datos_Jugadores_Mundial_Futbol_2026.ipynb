# Tratamiento de las inconsistencias

## Tratamiento de valores nulos

**1- Los valores nulos de las columnas tipo *float* son celdas vacías que serán rellenadas con el valor "0.0", el cual sería el valor que considero correspondiente como analista, en lugar de quedar vacío.**

```python


# Define las columnas que quieres rellenar
columns_to_fill = ['shots_on_target_pct', 'minutes', 'minutes_90s', 'goals', 'assists', 'goals_assists', 'goals_pens', 'pens_made', 'pens_att', 'cards_yellow', 'cards_red', 'goals_per90', 'assists_per90', 'goals_assists_per90', 'goals_pens_per90', 'goals_assists_pens_per90', 'shots', 'shots_on_target', 'shots_per90', 'shots_on_target_per90', 'goals_per_shot', 'goals_per_shot_on_target', 'minutes_per_game', 'minutes_pct', 'minutes_per_start', 'minutes_per_sub', 'points_per_game', 'on_goals_for', 'on_goals_against', 'plus_minus', 'plus_minus_per90', 'plus_minus_wowy', 'cards_yellow_red', 'fouls', 'fouled', 'offsides', 'crosses', 'interceptions', 'tackles_won', 'pens_won', 'pens_conceded', 'own_goals', 'gk_games', 'gk_games_starts', 'gk_minutes', 'gk_goals_against', 'gk_goals_against_per90', 'gk_shots_on_target_against', 'gk_saves', 'gk_save_pct', 'gk_wins', 'gk_ties', 'gk_losses', 'gk_clean_sheets', 'gk_clean_sheets_pct', 'gk_pens_att', 'gk_pens_allowed', 'gk_pens_saved', 'gk_pens_missed', 'gk_pens_save_pct' ]

# Rellena los valores nulos en esas columnas con 0.0
df[columns_to_fill] = df[columns_to_fill].fillna(0.0)

print("Valores nulos después de rellenar:")
display(df[columns_to_fill].isnull().sum())
```

**2- Los valores nulos de la columna tipo *object* "Club" serán rellenados con el valor "No Disponible", ya que no será de utlidad para el análisis buscar cada uno de los clubes de los jugadores que faltan completar.**

```python
df['club'] = df['club'].fillna("No Disponible")
```

```python
# Resultado luego de rellenar los valores nulos
df.info()
```

##Tratamiento de tipo de datos

**3- Algunas columnas que originalmente eran de tipo `float` y representaban conteos (por ejemplo, goles, asistencias, tarjetas) pueden convertirse a `int64` después de haber rellenado los valores nulos con 0.0, ya que estos valores son inherentemente enteros.**

```python
# Columnas a convertir de float a int64
columns_to_convert_to_int = [
    'minutes', 'goals', 'assists', 'goals_assists',
    'goals_pens', 'pens_made', 'pens_att', 'cards_yellow', 'cards_red',
    'shots', 'shots_on_target', 'minutes_per_game', 'minutes_per_start', 'minutes_per_sub',
    'on_goals_for', 'on_goals_against', 'plus_minus', 'cards_yellow_red',
    'fouls', 'fouled', 'offsides', 'crosses', 'interceptions', 'tackles_won',
    'pens_won', 'pens_conceded', 'own_goals',
    'gk_games', 'gk_games_starts', 'gk_minutes', 'gk_goals_against',
    'gk_shots_on_target_against', 'gk_saves', 'gk_wins', 'gk_ties', 'gk_losses',
    'gk_clean_sheets', 'gk_pens_att', 'gk_pens_allowed', 'gk_pens_saved',
    'gk_pens_missed'
]

for col in columns_to_convert_to_int:
    if col in df.columns:
        # Convertir a int64 después de asegurar que no hay nulos (ya rellenados con 0.0)
        df[col] = df[col].astype('int64')

print("Valores convertidos a int64:")
display(df[columns_to_convert_to_int].info())
```

## Reemplazo de valores en la columna 'position'

Para poder responder a las preguntas de investigación que tienen que ver con la posición en cancha de los jugadores, vamos a ver los valores únicos en la columna 'position'.

```python
print(df['position'].unique())
```

Ahora que conocemos los valores únicos, para poder realizar el análisis de la columna *position*, reemplazaremos los valores de jugadores que juegan en varias posiciones, dejando únicamente su posición original en la cancha.

```python
replacements = {
    'FW,MF': 'FW',
    'MF,FW': 'MF',
    'MF,DF': 'MF',
    'DF,MF': 'DF',
    'DF,FW': 'DF'
}

df['position'] = df['position'].replace(replacements)

print("Valores únicos de la columna 'position' después de los reemplazos:")
print(df['position'].unique())
```

## 2- Desempeño defensivo univariado: Análisis sobre interceptación de pases según posición de los jugadores en cancha

```python
plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x="interceptions", y="position", hue="position", palette="Set2", legend=False)
plt.title("Total de intercepciones de pases del rival según posición de los jugadores")
plt.xlabel("Cantidad de intercepciones")
plt.ylabel("Posición en cancha")
plt.show()
```

## 3- Análisis sobre la distribución grupal de las tareas ofensivas en los seleccionados finalistas del mundial, a través de la relación entre goles marcados (goals) y disparos intentados al arco rival (shots), según la posición de los jugadores (position) y sólo para los países finalistas del torneo (team_country)

```python
countries_to_filter = ['Argentina', 'Spain']
df_filtered = df[df['team_country'].isin(countries_to_filter)]

fig_filtered = px.scatter(df_filtered, x="goals", y="shots",
                         color="position",
                         facet_col="team_country",
                         hover_name="player",
                         title=f"Relación entre Goles marcados y disparos intentados, por posición, solo para {' y '.join(countries_to_filter)}")
fig_filtered.show()
```
